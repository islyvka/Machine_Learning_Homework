1. Навчіть на цих даних класифікатор kNN з параметрами за замовченням і виміряйте точність з допомогою AUROC на тренувальному та валідаційному наборах. Зробіть заключення про отриману модель: вона хороша/погана, чи є high bias/high variance?

```python
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier()
knn.fit(X_train, train_targets)
```

```python
train_probs = knn.predict_proba(X_train)
train_probs[:5]
```

```python
import matplotlib.pyplot as plt

plt.hist(train_probs[:,1])
plt.show()
```

```python
from sklearn.metrics import roc_auc_score

def predict_and_compute_auroc(inputs, targets, model, name=''):
  y_pred_proba = model.predict_proba(inputs)[:, 1]
  roc_auc = roc_auc_score(targets, y_pred_proba)

  print(f'AUROC for {name}: {roc_auc:.4f}')
```

```python
predict_and_compute_auroc(X_train, train_targets, knn, 'Training')
predict_and_compute_auroc(X_val, val_targets, knn, 'Validation')
```

**Спостереження**
- Модель з дефолтними параметрами вийшла заоверфітилась, підігналась під тестові дані та погано генералізує.

2. Використовуючи `GridSearchCV` знайдіть оптимальне значення параметра `n_neighbors` для класифікатора `kNN`. Псотавте крос валідацію на 5 фолдів.

  Після успішного завершення пошуку оптимального гіперпараметра
    - виведіть найкраще значення параметра
    - збережіть в окрему змінну `knn_best` найкращу модель, знайдену з `GridSearchCV`

```python
from sklearn.model_selection import GridSearchCV
import numpy as np

knn = KNeighborsClassifier()
params_knn = {'n_neighbors': np.arange(1, 25)}
knn_gs = GridSearchCV(knn, params_knn, cv=5)
knn_gs.fit(X_train, train_targets)
```

```python
knn_best = knn_gs.best_estimator_

print(knn_gs.best_params_, knn_gs.best_score_)
```

```python
knn_best = knn_gs.best_estimator_
```

```python
predict_and_compute_auroc(X_train, train_targets, knn_best, 'Training')
predict_and_compute_auroc(X_val, val_targets, knn_best, 'Validation')
```

**Спостереження**
- Найкращим значенням n_neighbors визначено 9. 
- Модель краща за попередню з дефолтними параметрами, але все ще є overfit.
- Щодо моделі DecisionTreeClassifier(max_depth=5), то на ній я отримала метрику 0.92 для валідаційного та тренувального наборів даних, тому вона залишається кращою.

    - виведіть найкращі значення параметра

    - оцініть якість передбачень  `dt_random_search_best` на тренувальній і валідаційній вибірці з допомогою AUROC.

```python
from sklearn.model_selection import RandomizedSearchCV
```

```python
%%time
dt_model = DecisionTreeClassifier(random_state=42)
params_dt = {
    'criterion': ['gini', 'entropy'],
    'splitter': ['best', 'random'],
    'max_depth': np.arange(1, 20),
    'max_leaf_nodes': np.arange(2, 20),
    'min_samples_split': [2, 5, 10, 20],
    'min_samples_leaf': [1, 2, 4, 8],
    'max_features': [None, 'sqrt', 'log2']
}
dt_rgs = RandomizedSearchCV(dt_model, params_dt, n_iter=40, cv=3, scoring='roc_auc')
dt_rgs.fit(X_train, train_targets)
```

```python
# найкращі значення параметрів
dt_rgs.best_params_
```

```python
# збереження
dt_random_search_best = dt_rgs.best_estimator_
```

```python
# оцінка
predict_and_compute_auroc(X_train, train_targets, dt_random_search_best, 'Training')
predict_and_compute_auroc(X_val, val_targets, dt_random_search_best, 'Validation')
```

**Спостереження**
- Пошук з RandomizedSearchCV дав кращі результати ніж всі попередні моделі.
